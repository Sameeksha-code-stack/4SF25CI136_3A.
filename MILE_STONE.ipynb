```python
df.drop_duplicates(inplace=True)
```

```python
df = df.drop(['Unnamed: 0', 'id'], axis=1)
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
mean_value = df['Arrival Delay in Minutes'].mean()

print("Mean Arrival Delay:", mean_value)
```

```python
df['Arrival Delay in Minutes'] = df['Arrival Delay in Minutes'].fillna(mean_value)
```

```python
df.isnull().sum()
```

```python
df['Arrival Delay in Minutes'] = df['Arrival Delay in Minutes'].astype(float)

df['Departure Delay in Minutes'] = df['Departure Delay in Minutes'].astype(float)
```

```python
df.dtypes
```

```python
df['Total Delay'] = (
    df['Departure Delay in Minutes'] +
    df['Arrival Delay in Minutes']
)
```

```python
sns.countplot(data=df, x='satisfaction')

plt.title('Passenger Satisfaction')
plt.xlabel('Satisfaction')
plt.ylabel('Number of Passengers')

plt.show()
```

```python
sns.countplot(data=df, x='Gender')

plt.title('Passenger Distribution by Gender')
plt.xlabel('Gender')
plt.ylabel('Number of Passengers')

plt.show()
```

```python
sns.histplot(df['Flight Distance'], kde=True)
plt.title('Flight Distance Distribution')
plt.xlabel('Flight Distance')
plt.ylabel('Frequency')
```

```python
sns.histplot(df['Age'], kde=True)
plt.title('Age Distribution of Passengers')
plt.xlabel('Age')
plt.ylabel('Frequency')
```

```python
plt.figure(figsize=(8,5))

plt.hist(df['Age'], bins=20)

plt.title('Age Distribution of Passengers')
plt.xlabel('Age')
plt.ylabel('Frequency')

plt.show()
```

```python
plt.figure(figsize=(15,10))

sns.heatmap(
    df.select_dtypes(include='number').corr(),
    annot=False,
    cmap='coolwarm'
)

plt.title('Correlation Heatmap')

plt.show()
```

```python
sns.catplot(
    x='satisfaction',
    y='Flight Distance',
    data=df,
    kind='box'
)

plt.show()
```

```python
sns.scatterplot(
    data=df,
    x='Flight Distance',
    y='Age',
    hue='satisfaction',
    palette='viridis'
)

plt.title('Age vs Flight Distance based on Satisfaction')
plt.show()
```
